# How to update the XPATH in the code:


### Go to the exact webpage containing the error. Each page is separated within the underlying code to help you find where the error took place. Python will also tell you which line caused the error in the error message that pops up.
### On the webpage, right-click on the dropdown, button, or tab that needs to be clicked and press Inspect. This will take you to the HTML that makes up the website.
### In the Inspect tab, each line of HTML code corresponds to something on the site. Both will be highlighted when you scroll over the HTML code, so find the HTML that corresponds with your dropdown/button/tab.
### Right-click on the HTML line and select Copy XPATH or Copy Full XPATH. Often, either option will work.
### Paste it in the lines of code causing the error. Note that most XPATHs appear in multiple lines of code (typically one locating the object and one interacting with it).
